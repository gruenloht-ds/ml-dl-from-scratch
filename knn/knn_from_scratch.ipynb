```python
import numpy as np
import pandas as pd
```

```python
classification_path='../toydata/breast_cancer_wisconsin_diagnositc/wdbc.csv'
df = pd.read_csv(classification_path)
```

```python
new_df = df.drop(['id', 'diagnosis'], axis=1)
col_names = new_df.columns
data = new_df.values
y = df['diagnosis'].values
y_bin = (y == 'M').astype(int)
```

## KNN

Steps:
1. calculate distance between new point and all other points
2. take the average of the top K labels as the label for the new point

```python
class myKNN():

    def __init__(self, k, distance):
        self.k = k
        self.distance = distance

    def fit(self, x, y):
        self.data = data
        self.y = y
    
    def predict(self, x):

        x_dist = self.distance(x, self.data)
        idx = np.argsort(x_dist)
        preds = self.y[idx][:self.k]

        pred = np.mean(preds)
        return pred
        
```

```python
def myDist(x, y):
    return np.sqrt(np.sum((x - y)**2, axis=1))
```

```python
knn = myKNN(5, myDist)
```

```python
knn.fit(data, y_bin)
```

```python
knn.predict(data[500])
```
